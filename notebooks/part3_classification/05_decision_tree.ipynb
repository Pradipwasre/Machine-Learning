# Decision Trees – Notes

## Introduction
- Decision Trees are a simple yet powerful machine learning algorithm.
- They are the foundation for advanced models like **Random Forest**.
- Can be used for both **Classification** (categorical outputs) and **Regression** (continuous outputs).

---

## Core Algorithms
- **ID3 Algorithm** → based on *Entropy*.
- **CART Algorithm** → based on *Gini Index*.
- Both rely on **Information Gain** to decide the best splits.

---

## Key Concepts
- **Purity of Splits** → how homogeneous the groups are after splitting.
- **Information Gain (IG)** → reduction in impurity after a split.
- Splits resemble **binary tree structures** (Yes/No, True/False).

---

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Import the dataset from CSV file
dataset = pd.read_csv('Social_Network_Ads.csv')

# Separate features (X) and target (y)
# iloc[:, :-1] → all rows, all columns except the last one (features)
X = dataset.iloc[:, :-1].values

# iloc[:, -1] → all rows, only the last column (target variable)
y = dataset.iloc[:, -1].values
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 0)
```

```python
# Import the StandardScaler class from sklearn
from sklearn.preprocessing import StandardScaler

# Create a StandardScaler object
sc = StandardScaler()

# Fit the scaler on the training data and transform it
# This calculates mean and standard deviation from X_train
# and scales the values so they have mean=0 and variance=1
X_train = sc.fit_transform(X_train)

# Transform the test data using the same scaler
# Important: we only transform here, not fit again,
# so the test data is scaled using training statistics
X_test = sc.transform(X_test)
```

#  Why Use Entropy in Decision Trees?

### What is Entropy?
- Entropy is a measure of **disorder or impurity** in the data.
- If a node contains samples from only one class → Entropy = 0 (perfectly pure).
- If a node contains a mix of classes → Entropy > 0 (impure).

---

### How Entropy Helps the Algorithm
- At each split, the decision tree checks **all features** and calculates how much entropy would reduce if we split on that feature.
- The feature that gives the **highest reduction in entropy** (called **Information Gain**) is chosen for the split.
- This ensures that each step makes the groups **purer** and classification more accurate.

---

### Simplest Way to Think About It
- Imagine sorting marbles:
  - If a box has only red marbles → no confusion (entropy = 0).
  - If a box has half red and half green → maximum confusion (entropy is high).
- The tree keeps splitting until each box (leaf node) is as pure as possible.

---

*Entropy tells the tree how “mixed up” the data is. The tree always splits on the feature that reduces this mix the most, so each branch becomes clearer and more class‑pure.*


```python
# Import the DecisionTreeClassifier class from sklearn
from sklearn.tree import DecisionTreeClassifier

# Create a Decision Tree classifier object
# criterion='entropy' → tells the tree to use Entropy (information gain) to decide splits
# random_state=0 → ensures reproducibility (same results every run)
classifier = DecisionTreeClassifier(criterion='entropy', random_state=0)

# Fit the classifier on the training data
# This trains the decision tree using X_train (features) and y_train (labels)
classifier.fit(X_train, y_train)
```

-  random_state is like setting a "seed" for randomness
-  Many ML algorithms use random steps internally (like shuffling data or choosing splits)
-  If you don't fix the random_state, every run can give slightly different results
- By setting random_state=0 (or any fixed number), you make the randomness reproducible
- → This means you and other person will always get the same tree structure and results
- → Useful for teaching, debugging, and comparing models fairly

```python
# Predict the class for a new data point
# Example: Age = 30, Estimated Salary = 87000
# We must scale the input using the same StandardScaler (sc) before prediction
print(classifier.predict(sc.transform([[30, 87000]])))
```

```python
# Predict the target values for the test set
y_pred = classifier.predict(X_test)

# Print predictions side by side with actual test labels
# np.concatenate → joins arrays together
# y_pred.reshape(len(y_pred),1) → reshape predictions into a column
# y_test.reshape(len(y_test),1) → reshape actual values into a column
# axis=1 → concatenate them horizontally (side by side)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)), axis=1))
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score

# Build the confusion matrix (shows correct vs incorrect predictions)
cm = confusion_matrix(y_test, y_pred)
print(cm)

# Calculate and return the accuracy score (overall % correct)
accuracy_score(y_test, y_pred)
```
