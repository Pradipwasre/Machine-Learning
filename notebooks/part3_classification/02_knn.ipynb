```python
# K-Nearest Neighbors (K-NN)
# This section sets up the environment for using the KNN algorithm.

# Importing the libraries
import numpy as np              # NumPy is used for numerical operations and handling arrays
import matplotlib.pyplot as plt # Matplotlib is used for plotting graphs and visualizing data
import pandas as pd             # Pandas is used for handling datasets in tabular form (rows and columns)
```

```python
# Importing the dataset
dataset = pd.read_csv('Social_Network_Ads.csv')  
# Reads the CSV file into a Pandas DataFrame. 
# Each row represents one user’s record (Age, Estimated Salary, Purchased).

X = dataset.iloc[:, :-1].values  
# Selects all columns except the last one as features (Age and Salary). 
# .iloc[:, :-1] means “all rows, all columns except the last.”

y = dataset.iloc[:, -1].values  
# Selects the last column as the target variable (Purchased: 0 = No, 1 = Yes).
# .iloc[:, -1] means “all rows, only the last column.”
```

```python
# Splitting the dataset into the Training set and Test set
from sklearn.model_selection import train_test_split  

# train_test_split is a function that divides the dataset into two parts:
# - Training set: used to teach the model
# - Test set: used to evaluate how well the model performs on unseen data

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size = 0.25, random_state = 0
)
# test_size = 0.25 → 25% of the data goes into the test set, 75% into the training set
# random_state = 0 → ensures reproducibility (same split every time you run the code)
```

```python
# Feature Scaling
from sklearn.preprocessing import StandardScaler  

# StandardScaler standardizes features by removing the mean and scaling to unit variance.
# This ensures that Age and Salary are on the same scale, so distance calculations in KNN are fair.

sc = StandardScaler()  
# Creates a scaler object that will learn the mean and standard deviation from the training set.

X_train = sc.fit_transform(X_train)  
# Fits the scaler on the training data (learns mean & std) and transforms it.
# Result: Age and Salary values are standardized (mean = 0, std = 1).

X_test = sc.transform(X_test)  
# Uses the same scaler (from training set) to transform the test data.
# Important: We never fit again on test data, to avoid data leakage.

print(X_train)  # Shows the scaled training features
print(X_test)   # Shows the scaled test features
```

```python
# Training the K-NN model on the Training set
from sklearn.neighbors import KNeighborsClassifier  

# KNeighborsClassifier is the scikit-learn class for KNN.
# It classifies new data points based on the majority class of their nearest neighbors.

classifier = KNeighborsClassifier(
    n_neighbors = 5,   # K = 5 → the algorithm will look at the 5 closest neighbors
    metric = 'minkowski', # Minkowski is a general distance metric
    p = 2              # When p = 2, Minkowski becomes Euclidean distance (straight-line distance)
)

classifier.fit(X_train, y_train)  
# Fits the classifier on the training data.
# The model now "stores" the training points and will use them to predict new cases.
```

```python
# Predicting a new result
print(classifier.predict(sc.transform([[30, 87000]])))
# Explanation:
# - We want to predict whether a person aged 30 with a salary of 87,000 will purchase or not.
# - [[30, 87000]] is the new data point (Age = 30, Salary = 87,000).
# - sc.transform(...) scales this new data point using the same scaler fitted on training data.
#   This ensures the new input is on the same scale as the training set.
# - classifier.predict(...) applies the KNN model to predict the outcome.
# - Output will be [0] or [1]:
#     0 → Not Purchased
#     1 → Purchased
```

```python
# Predicting the Test set results
y_pred = classifier.predict(X_test)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

```python
# Predicting a new result
print(classifier.predict(sc.transform([[30, 87000]])))
# Explanation:
# - We want to predict whether a person aged 30 with a salary of 87,000 will purchase or not.
# - [[30, 87000]] is the new data point (Age = 30, Salary = 87,000).
# - sc.transform(...) scales this new data point using the same scaler fitted on training data.
#   This ensures the new input is on the same scale as the training set.
# - classifier.predict(...) applies the KNN model to predict the outcome.
# - Output will be [0] or [1]:
#     0 → Not Purchased
#     1 → Purchased


# Making the Confusion Matrix
from sklearn.metrics import confusion_matrix, accuracy_score  

# confusion_matrix compares the actual test labels (y_test) with the predicted labels (y_pred).
# It shows how many predictions were correct (True Positives & True Negatives)
# and how many were wrong (False Positives & False Negatives).

cm = confusion_matrix(y_test, y_pred)  
print(cm)  
# Prints the confusion matrix as a 2x2 table:
# [[TN, FP],
#  [FN, TP]]

accuracy_score(y_test, y_pred)  
# Calculates the overall accuracy of the model:
# Accuracy = (Correct Predictions) / (Total Predictions)
```

- TN = 64 → 64 people correctly predicted as Not Purchased

- FP = 4 → 4 people wrongly predicted as Purchased when they actually didn’t

- FN = 3 → 3 people wrongly predicted as Not Purchased when they actually did purchase

- TP = 29 → 29 people correctly predicted as Purchased

# 🧭 Pro Tips for Understanding K-Nearest Neighbors (KNN)

KNN is one of the simplest yet powerful algorithms in machine learning.  
Here’s how to make sense of the entire workflow we coded step by step:

---

## 1. Importing Libraries
- **NumPy** → Math toolbox (arrays, calculations).
- **Matplotlib** → Drawing board (plots, charts).
- **Pandas** → Spreadsheet manager (datasets in rows & columns).

👉 *Tip:* Always know which library does what — it helps you debug faster.

---

## 2. Loading the Dataset
- `X` = Features (Age, Salary).  
- `y` = Target (Purchased or Not).  

👉 *Tip:* Think of `X` as the **questions** and `y` as the **answers**.

---

## 3. Splitting into Training & Test Sets
- Training set → Model learns patterns.  
- Test set → Model proves itself.  

👉 *Tip:* This is like practicing before an exam and then sitting for the real test.

---

## 4. Feature Scaling
- KNN relies on **distance**.  
- If features are on different scales (Age in years vs. Salary in thousands), one feature dominates.  
- Scaling ensures fair comparison.  

👉 *Tip:* Always scale before using KNN, otherwise results can be misleading.

---

## 5. Training the KNN Model
- `n_neighbors = 5` → Look at 5 closest points.  
- `metric = 'minkowski', p = 2` → Euclidean distance.  
- KNN doesn’t build a formula, it just **remembers the training data**.

👉 *Tip:* Choosing `K` is crucial:
- Small K → sensitive to noise.  
- Large K → smoother but may ignore local patterns.

---

## 6. Predicting New Results
- Input a new data point (Age = 30, Salary = 87,000).  
- Scale it → Compare with training data → Majority vote decides.  

 *Tip:* Always transform new inputs with the same scaler used on training data.

---

## 7. Confusion Matrix & Accuracy
- Confusion Matrix shows **types of predictions**:
  - TN → Correctly predicted Not Purchased.
  - FP → Wrongly predicted Purchased.
  - FN → Wrongly predicted Not Purchased.
  - TP → Correctly predicted Purchased.
- Accuracy = (TN + TP) / Total predictions.

 *Tip:* Don’t rely only on accuracy. Look at **Precision** and **Recall** to understand the model’s strengths and weaknesses.

---

##  Big Picture
- KNN is like asking your neighbors for advice, the majority opinion decides the prediction.
- It’s simple, intuitive, and great for small datasets.
- Scaling and choosing the right `K` are the keys to success.

---

**Trainer one-liner for students:**  
*KNN predicts by looking at the closest neighbors, like asking your friends for advice. The majority vote decides the outcome.*


#  Precision vs Recall in KNN (Which to Pick?)

When evaluating a KNN model, we often ask: should we care more about **Precision** or **Recall**?  
The answer depends on the business problem and the cost of mistakes.

---

## Definitions
- **Precision** = TP / (TP + FP)  
  → Of all the people we predicted as "Purchased," how many actually purchased?  
  → Focuses on **avoiding false alarms**.

- **Recall** = TP / (TP + FN)  
  → Of all the people who actually purchased, how many did we correctly predict?  
  → Focuses on **catching as many positives as possible**.

---

## In Our Dataset (Customer Purchase Prediction)
- **False Positive (FP):** Predicting someone will purchase when they won’t.  
  → Cost: Wasted marketing effort (sending offers to uninterested customers).  

- **False Negative (FN):** Predicting someone won’t purchase when they actually will.  
  → Cost: Missed opportunity (losing a potential buyer).  

---

## Which Matters More?
- In marketing and sales, **Recall is usually more important**.  
  - Missing a buyer (FN) means losing revenue.  
  - A few false alarms (FP) are acceptable because sending an extra offer is cheaper than missing a sale.  

- In fraud detection or spam filtering, **Precision is more important**.  
  - False alarms (FP) are costly because they wrongly flag genuine customers or emails.  

---

## For Our Case (Social Network Ads → Purchase Prediction)
- **We prioritize Recall**:  
  - Better to catch more potential buyers, even if we sometimes predict a few non-buyers as buyers.  
  - This ensures we maximize sales opportunities.  

---

### Pro Tip for Students
*Think about the real-world cost of mistakes. If missing positives is worse, choose Recall. If false alarms are worse, choose Precision. Accuracy alone is not enough  always ask which metric aligns with the business goal.*
